```python
from dotenv import load_dotenv
import openai
load_dotenv()
client = openai.OpenAI()
```

```python
with open("data/ch06_quiz.txt", "r") as f:
    ori_text = f.read()
```

```python
print(type(ori_text))
```

```python
from openai import OpenAI

response = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "system", "content": "다음 글을 50자 이내로 요약해줘."},
        {"role": "user", "content": ori_text}
    ]
)

print(response.choices[0].message.content.strip())
```

```python
speech_response = client.audio.speech.with_streaming_response.create(
    model="tts-1",
    voice="nova",
    input=response
)

# MP3 파일로 저장
with open("ch06_quiz.mp3", "wb") as f:
    for chunk in speech_response.iter_bytes():
        f.write(chunk)

print("음성 파일 저장 완료: ch06_quiz.mp3")
```
