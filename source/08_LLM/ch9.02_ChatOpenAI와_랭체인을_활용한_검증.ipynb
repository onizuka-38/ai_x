# 2. LLM 답변 생성
## 2.1 openAI SDK 를 선택
- 세밀한 제어가 필요할 때, 성능 최적화가 중요할 때

```python
from openai import OpenAI
client = OpenAI()
response = client.chat.completions.create(
    model="gpt-4.1-nano",
    messages =[
        {"role": "system", "content": "너는 친절하게 대답해 주는 비서야"},
        {"role": "user", "content": "2020년 월드 시리즈 우승팀은 어디야?"},
    ]
)
print(response.choices[0].message.content)
```

## 2.2 LangChain을 선택
- 내부적으로 자동 메세지 형태를 변환해 줌

```python
from langchain_openai import ChatOpenAI
# .env 파일에 OPENAI_API_KEY가 설정되어 있어야 합니다.
llm = ChatOpenAI(model="gpt-4.1-nano")
ai_message = llm.invoke("2020년 월드 시리즈 우승팀은 어디야?")
print(ai_message.content)
```
