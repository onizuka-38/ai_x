## 1-1 분류분석이란?

## 1-2 샘플용 데이터 셋 접근
### load 계열(패키지 내에 데이터 있음)
    - load_iris(), load_boston(), load_diabetes()...
### fetch 계열(패키지 내에 데이터가 없고, 인터넷 연결시 다운)
    - fetch_openml(), fetch_rov1(로이터뉴스 말뭉치)
### make 계열(가상의 데이터 생성)
    - make_classification() : 분류분석용 데이터 생성
    - make_regression() : 회귀분석용 데이터 생성
    - make_blobs() : 군집분석용 데이터 생성(타겟변수 x)

```python
# 패키지 호환성 문제 해결을 위해 기본 Python으로 간단한 분류 데이터 생성
import random

# 간단한 2차원 분류 데이터 생성 (iris 데이터 유사)
random.seed(42)

# 클래스 0: 작은 값들
class_0_data = []
for i in range(50):
    x1 = random.uniform(1.0, 3.0)  # 꽃받침 길이
    x2 = random.uniform(0.5, 1.5)  # 꽃받침 너비
    x3 = random.uniform(1.0, 2.0)  # 꽃잎 길이  
    x4 = random.uniform(0.1, 0.3)  # 꽃잎 너비
    class_0_data.append([x1, x2, x3, x4, 0])

# 클래스 1: 중간 값들
class_1_data = []
for i in range(50):
    x1 = random.uniform(4.0, 6.0)
    x2 = random.uniform(2.0, 3.5)
    x3 = random.uniform(3.0, 5.0)
    x4 = random.uniform(1.0, 1.8)
    class_1_data.append([x1, x2, x3, x4, 1])

# 클래스 2: 큰 값들
class_2_data = []
for i in range(50):
    x1 = random.uniform(5.5, 8.0)
    x2 = random.uniform(2.5, 4.0)
    x3 = random.uniform(4.5, 7.0)
    x4 = random.uniform(1.5, 2.5)
    class_2_data.append([x1, x2, x3, x4, 2])

# 모든 데이터 합치기
all_data = class_0_data + class_1_data + class_2_data
random.shuffle(all_data)

# X, y 분리
X = [[row[0], row[1], row[2], row[3]] for row in all_data]
y = [row[4] for row in all_data]

# 훈련/테스트 분할
train_size = int(0.8 * len(X))
X_train = X[:train_size]
y_train = y[:train_size]
X_test = X[train_size:]
y_test = y[train_size:]

print(f"데이터 생성 완료!")
print(f"훈련 데이터: {len(X_train)}개")
print(f"테스트 데이터: {len(X_test)}개")
print(f"클래스 분포: 0={y.count(0)}, 1={y.count(1)}, 2={y.count(2)}")
print(f"첫 번째 샘플: {X[0]} -> 클래스 {y[0]}")
```

```python
# matplotlib 호환성 문제로 인해 데이터 정보를 텍스트로 출력
print("=== 첫 번째 샘플 정보 ===")
print(f"특성값: {X_train[0]}")
print(f"클래스: {y_train[0]}")
print()

print("=== 각 클래스별 샘플 하나씩 ===")
for class_num in [0, 1, 2]:
    for i, label in enumerate(y_train):
        if label == class_num:
            print(f"클래스 {class_num}: {X_train[i]}")
            break

print()
print("=== 데이터 통계 ===")
print(f"총 훈련 샘플: {len(X_train)}")
print(f"특성 개수: {len(X_train[0])}")
print("특성별 평균값:")
for j in range(len(X_train[0])):
    avg = sum(row[j] for row in X_train) / len(X_train)
    print(f"  특성 {j+1}: {avg:.2f}")
```

# 2절. 분류분석의 종류

## 2-1 확률적 모형
    - 주어진 데이터에 각 클래스가 정답일 확률을 계산하는 모형
    - predict_proba() : 각 분류 클래스별 확률을 알려줌(일부의 판별함수 모형에 있긴 함)
    - predict_log_proba() : 확률의 로그값을 알려줌

### 1) 확률적 생성 모형
    - 각 클래스별 특정 데이터 확률분포 추정(베이즈 정리를 사용), 소량의 데이터로도 작동
    - ex. Quadratic Discriminant Analysis, 나이브 베이지안

### 2) 확률적 판별 모형
    - 확률을 계산하는 함수 이용
    - ex. 의사결정나무, 로지스틱 회귀

## 2-2 판별함수 모형
    - 주어진 데이터를 클래스에 따라 서로 다른 영역으로 나누는 경계면을 찾은 후,
    - 이 경계면으로부터 어느 위치에 있는지를 계산하는 판별함수를 이용
    - decision_function()
    - ex. 퍼셉트론, 서포트벡터머신, MLP(신경망)

=> RandomRorest, LGBM, XGBoost, CatBoost 등의 앙상블 모형이 더 많이 쓰임

```python
# 분류모형의 시각화
import io
from sklearn.tree import export_graphviz
import pydot
from IPython.core.display import Image

def draw_decision_tree(model, feature_names):
    dot_buf = io.StringIO()
    export_graphviz(model, out_file=dot_buf, feature_names=feature_names)
    graph = pydot.graph_from_dot_data(dot_buf.getvalue())[0]
    image = graph.create_png()
    return Image(image)
draw_decision_tree(dt_model, feature_names=feature_names)
```

```python
import matplotlib as mpl
import numpy as np
import matplotlib.pyplot as plt
def plot_decision_regions(X, y, model, title):
    species = ['setosa', 'versicolor', 'virginica']
    resolution = 0.01
    markers = ('s', '^', 'o')
    colors = ('red', 'blue', 'lightgreen')
    cmap = mpl.colors.ListedColormap(colors)

    x1_min, x1_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    x2_min, x2_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx1, xx2 = np.meshgrid(np.arange(x1_min, x1_max, resolution),
                           np.arange(x2_min, x2_max, resolution))
    Z = model.predict(
        np.array([xx1.ravel(), xx2.ravel()]).T).reshape(xx1.shape)

    plt.contour(xx1, xx2, Z, cmap=mpl.colors.ListedColormap(['k']))
    plt.contourf(xx1, xx2, Z, alpha=0.4, cmap=cmap)
    plt.xlim(xx1.min(), xx1.max())
    plt.ylim(xx2.min(), xx2.max())

    for idx, cl in enumerate(np.unique(y)):
        plt.scatter(x=X[y == cl, 0], y=X[y == cl, 1], alpha=0.8,
                    c=[cmap(idx)], marker=markers[idx], s=80, label=species[cl])

    plt.xlabel(data.feature_names[2])
    plt.ylabel(data.feature_names[3])
    plt.legend(loc='upper left')
    plt.title(title)

    return Z
plot_decision_regions(X, y, dt_model, "Depth1")
plt.show()
```

## 6-3 ROC커브를 이용한 성능 비교
- ROC그래프의 가로축 위양성율(fprate : False Positive rate), 세로축 racall값

```python
# sklearn 호환성 문제로 인해 기본 Python으로 불균형 분류 데이터 생성
import random

random.seed(5)

# 불균형 데이터 생성 (95% vs 5% 비율)
X_imbalanced = []
y_imbalanced = []

# 클래스 0 (95% - 950개)
for i in range(950):
    x1 = random.uniform(0, 5) + random.gauss(0, 0.5)
    x2 = random.uniform(0, 5) + random.gauss(0, 0.5)
    X_imbalanced.append([x1, x2])
    y_imbalanced.append(0)

# 클래스 1 (5% - 50개)  
for i in range(50):
    x1 = random.uniform(3, 8) + random.gauss(0, 0.5)
    x2 = random.uniform(3, 8) + random.gauss(0, 0.5)
    X_imbalanced.append([x1, x2])
    y_imbalanced.append(1)

# 데이터 섞기
combined = list(zip(X_imbalanced, y_imbalanced))
random.shuffle(combined)
X_imbalanced, y_imbalanced = zip(*combined)
X_imbalanced = list(X_imbalanced)
y_imbalanced = list(y_imbalanced)

print(f"불균형 데이터 생성 완료!")
print(f"총 샘플 수: {len(X_imbalanced)}")
print(f"클래스 0: {y_imbalanced.count(0)}개 ({y_imbalanced.count(0)/len(y_imbalanced)*100:.1f}%)")
print(f"클래스 1: {y_imbalanced.count(1)}개 ({y_imbalanced.count(1)/len(y_imbalanced)*100:.1f}%)")
print(f"첫 번째 샘플: {X_imbalanced[0]} -> 클래스 {y_imbalanced[0]}")
```

```python
# 기본 Python으로 간단한 분류 모델 구현
import math

class SimpleClassifier:
    """간단한 거리 기반 분류기"""
    
    def __init__(self):
        self.train_X = None
        self.train_y = None
        
    def fit(self, X, y):
        """훈련 데이터 저장"""
        self.train_X = X.copy()
        self.train_y = y.copy()
        
    def predict_proba(self, X):
        """각 샘플의 클래스별 확률 예측 (k-NN 기반)"""
        probabilities = []
        
        for test_point in X:
            # 모든 훈련 샘플과의 거리 계산
            distances = []
            for i, train_point in enumerate(self.train_X):
                dist = math.sqrt(sum((a-b)**2 for a, b in zip(test_point, train_point)))
                distances.append((dist, self.train_y[i]))
            
            # 가장 가까운 5개 이웃 찾기
            distances.sort()
            k_neighbors = distances[:5]
            
            # 클래스별 투표
            class_0_votes = sum(1 for _, label in k_neighbors if label == 0)
            class_1_votes = sum(1 for _, label in k_neighbors if label == 1)
            
            # 확률 계산
            total_votes = class_0_votes + class_1_votes
            prob_0 = class_0_votes / total_votes
            prob_1 = class_1_votes / total_votes
            
            probabilities.append([prob_0, prob_1])
            
        return probabilities
    
    def predict(self, X):
        """클래스 예측"""
        probas = self.predict_proba(X)
        return [0 if prob[0] > prob[1] else 1 for prob in probas]

# 모델 훈련
model = SimpleClassifier()
model.fit(X_imbalanced, y_imbalanced)

# 예측
y_pred = model.predict(X_imbalanced)
y_proba = model.predict_proba(X_imbalanced)

print("=== 분류 모델 훈련 완료 ===")
print(f"예측 완료: {len(y_pred)}개 샘플")
print(f"첫 5개 예측 확률:")
for i in range(5):
    print(f"샘플 {i+1}: 클래스0={y_proba[i][0]:.3f}, 클래스1={y_proba[i][1]:.3f} -> 예측={y_pred[i]}")
```

```python
# 성능 평가 메트릭 구현
def confusion_matrix(y_true, y_pred):
    """혼동행렬 계산"""
    tp = sum(1 for true, pred in zip(y_true, y_pred) if true == 1 and pred == 1)
    tn = sum(1 for true, pred in zip(y_true, y_pred) if true == 0 and pred == 0)
    fp = sum(1 for true, pred in zip(y_true, y_pred) if true == 0 and pred == 1)
    fn = sum(1 for true, pred in zip(y_true, y_pred) if true == 1 and pred == 0)
    return {'TP': tp, 'TN': tn, 'FP': fp, 'FN': fn}

def calculate_metrics(y_true, y_pred):
    """성능 지표 계산"""
    cm = confusion_matrix(y_true, y_pred)
    tp, tn, fp, fn = cm['TP'], cm['TN'], cm['FP'], cm['FN']
    
    # 정확도 (Accuracy)
    accuracy = (tp + tn) / (tp + tn + fp + fn)
    
    # 정밀도 (Precision)
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    
    # 재현율 (Recall/Sensitivity)
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    
    # 특이도 (Specificity)
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
    
    # F1 Score
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    
    return {
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'specificity': specificity,
        'f1_score': f1,
        'confusion_matrix': cm
    }

# 성능 평가
metrics = calculate_metrics(y_imbalanced, y_pred)

print("=== 모델 성능 평가 ===")
print(f"정확도 (Accuracy): {metrics['accuracy']:.3f}")
print(f"정밀도 (Precision): {metrics['precision']:.3f}")
print(f"재현율 (Recall): {metrics['recall']:.3f}")
print(f"특이도 (Specificity): {metrics['specificity']:.3f}")
print(f"F1 Score: {metrics['f1_score']:.3f}")
print()
print("=== 혼동행렬 ===")
cm = metrics['confusion_matrix']
print(f"               예측")
print(f"실제    0     1")
print(f"  0   {cm['TN']:3d}  {cm['FP']:3d}")
print(f"  1   {cm['FN']:3d}  {cm['TP']:3d}")
```

```python
# ROC 커브와 AUC 계산
def calculate_roc_curve(y_true, y_proba):
    """ROC 커브 계산"""
    # 확률 임계값들 생성 (0.0부터 1.0까지)
    thresholds = [i/100 for i in range(101)]
    
    fpr_list = []  # False Positive Rate
    tpr_list = []  # True Positive Rate (Recall)
    
    for threshold in thresholds:
        # 임계값에 따른 예측
        y_pred_threshold = [1 if prob[1] >= threshold else 0 for prob in y_proba]
        
        # 혼동행렬 계산
        cm = confusion_matrix(y_true, y_pred_threshold)
        tp, tn, fp, fn = cm['TP'], cm['TN'], cm['FP'], cm['FN']
        
        # FPR = FP / (FP + TN)
        fpr = fp / (fp + tn) if (fp + tn) > 0 else 0
        
        # TPR = TP / (TP + FN) = Recall
        tpr = tp / (tp + fn) if (tp + fn) > 0 else 0
        
        fpr_list.append(fpr)
        tpr_list.append(tpr)
    
    return fpr_list, tpr_list, thresholds

def calculate_auc(fpr_list, tpr_list):
    """AUC 계산 (사다리꼴 공식)"""
    auc = 0
    for i in range(1, len(fpr_list)):
        # 사다리꼴 넓이 = (상저 + 하저) * 높이 / 2
        width = fpr_list[i] - fpr_list[i-1]
        height = (tpr_list[i] + tpr_list[i-1]) / 2
        auc += width * height
    return auc

# ROC 커브 계산
fpr, tpr, thresholds = calculate_roc_curve(y_imbalanced, y_proba)
auc_score = calculate_auc(fpr, tpr)

print("=== ROC 커브 분석 ===")
print(f"AUC Score: {auc_score:.3f}")
print()
print("ROC 커브 주요 지점들:")
print("Threshold  FPR    TPR")
print("-" * 25)
for i in range(0, len(thresholds), 10):  # 10개씩 건너뛰며 출력
    print(f"{thresholds[i]:8.2f}  {fpr[i]:.3f}  {tpr[i]:.3f}")

# ROC 커브의 일부 구간을 텍스트로 시각화
print()
print("=== ROC 커브 시각화 (텍스트) ===")
print("TPR")
print("1.0 |" + "".join("*" if any(abs(tpr[j] - (10-i)/10) < 0.05 and abs(fpr[j] - k/20) < 0.025 
                              for j in range(len(tpr))) else " " 
                        for k in range(21)))
print("0.5 |" + "".join("*" if any(abs(tpr[j] - 0.5) < 0.05 and abs(fpr[j] - k/20) < 0.025 
                              for j in range(len(tpr))) else " " 
                        for k in range(21)))
print("0.0 |" + "-" * 21)
print("    0.0" + " " * 15 + "1.0 FPR")
```

```python
# AUC 계산 수정 (올바른 방법)
def calculate_auc_corrected(fpr_list, tpr_list):
    """AUC 계산 (올바른 사다리꼴 공식)"""
    # FPR을 기준으로 정렬
    points = list(zip(fpr_list, tpr_list))
    points.sort()  # FPR 기준으로 정렬
    
    auc = 0
    for i in range(1, len(points)):
        # x축 간격 (FPR 차이)
        dx = points[i][0] - points[i-1][0]
        # y축 평균 (TPR 평균)
        avg_y = (points[i][1] + points[i-1][1]) / 2
        # 넓이 누적
        auc += dx * avg_y
    
    return auc

# 수정된 AUC 계산
auc_corrected = calculate_auc_corrected(fpr, tpr)

print("=== 수정된 ROC 분석 ===")
print(f"올바른 AUC Score: {auc_corrected:.3f}")
print()

# ROC 포인트들을 정렬하여 출력
roc_points = list(zip(fpr, tpr, thresholds))
roc_points.sort()  # FPR 기준 정렬

print("정렬된 ROC 포인트들 (FPR 기준):")
print("   FPR    TPR  Threshold")
print("-" * 28)
for i in range(0, len(roc_points), 10):
    fpr_val, tpr_val, thresh = roc_points[i]
    print(f"{fpr_val:6.3f} {tpr_val:6.3f}     {thresh:.2f}")

print()
print("=== 모델 성능 해석 ===")
if auc_corrected > 0.9:
    performance = "우수함"
elif auc_corrected > 0.8:
    performance = "좋음"  
elif auc_corrected > 0.7:
    performance = "보통"
elif auc_corrected > 0.6:
    performance = "나쁨"
else:
    performance = "매우 나쁨"

print(f"AUC {auc_corrected:.3f}는 '{performance}' 성능을 나타냅니다.")
print(f"이는 랜덤 분류기(AUC=0.5)보다 {auc_corrected-0.5:.3f}만큼 좋습니다.")
```
