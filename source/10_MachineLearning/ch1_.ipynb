# 2절. 데이터 탐색
## 2.1 변수의 종류
- 데이터 속성에 따라 명목식(남,여), 순서식(서열, 5점척도), 구간식(순서, 간격의미) 비율식(연속형, 영존재)으로 데이터 구분
- 변수의 타입에 따라 연속형, 이산형, 범주형
- 데이터 품질에 따라 특이값(비정상치, 이상치), 결측값

## 2.2 통계요약; 중심위치 측도(평균, 중앙값=중위수, 최빈값), 산포측도(왜도, 점도)

```python
import numpy as np
from scipy.stats import kurtosis, skew
```

```python
# 평균
data = np.random.normal(loc=0, scale=1.0, size=10000000)
data.shape
```

```python
# 분포도(히스토그램)
import matplotlib.pyplot as plt
plt.figure(figsize=(5,1))
plt.hist(data, bins=150)
plt.show()
```

```python
print('평균 :', np.mean(data))
print('중앙값 :', np.median(data))
print('분산 :', np.var(data))
print('표준편차 :', np.std(data))
```

```python
# 왜도(skewness) : 0(대칭), 음수(왼쪽으로 긴꼬리 분포)
print('왜도(치우친 정도) :', skew(data))
```
