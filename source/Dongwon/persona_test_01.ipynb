```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# --- 설정값 ---
NUM_CUSTOMERS = 1000
NUM_ORDERS = 5000
START_DATE = datetime(2023, 8, 1)
END_DATE = datetime(2025, 7, 31)

# 상품 카테고리와 가격 범위 정의 (가중치를 두어 현실성 부여)
product_categories = {
    '건강기능식품': {'price_range': (20000, 80000), 'weight': 0.4},
    '가공식품': {'price_range': (5000, 30000), 'weight': 0.3},
    '생활용품': {'price_range': (10000, 50000), 'weight': 0.2},
    '뷰티': {'price_range': (15000, 60000), 'weight': 0.1}
}

# --- 데이터 생성 ---
print("가상 데이터 생성을 시작합니다...")

# 고객 ID 생성
customer_ids = [f"C{str(i).zfill(4)}" for i in range(1, NUM_CUSTOMERS + 1)]

# 주문 데이터 리스트 생성
orders_list = []
categories = list(product_categories.keys())
weights = [product_categories[cat]['weight'] for cat in categories]

for i in range(1, NUM_ORDERS + 1):
    # 주문 정보 랜덤 선택
    order_id = f"O{str(i).zfill(5)}"
    customer_id = random.choice(customer_ids)
    
    # 날짜 랜덤 생성
    random_days = random.randint(0, (END_DATE - START_DATE).days)
    order_date = START_DATE + timedelta(days=random_days)
    
    # 상품 카테고리 및 가격 랜덤 생성 (가중치 적용)
    product_category = random.choices(categories, weights=weights, k=1)[0]
    min_price, max_price = product_categories[product_category]['price_range']
    price = random.randint(min_price // 1000, max_price // 1000) * 1000
    
    # 상품 ID는 카테고리와 주문번호 조합으로 간단히 생성
    product_id = f"P-{product_category[:2]}-{str(i).zfill(5)}"

    orders_list.append([
        order_id,
        customer_id,
        order_date,
        product_id,
        product_category,
        price
    ])

# 데이터프레임으로 변환
df_orders = pd.DataFrame(
    orders_list,
    columns=['OrderID', 'CustomerID', 'OrderDate', 'ProductID', 'ProductCategory', 'Price']
)

# 데이터 타입 변환
df_orders['OrderDate'] = pd.to_datetime(df_orders['OrderDate'])
df_orders['Price'] = df_orders['Price'].astype(int)

# --- CSV 파일로 저장 ---
output_filename = 'customer_orders.csv'
df_orders.to_csv(output_filename, index=False, encoding='utf-8-sig')

print("-" * 30)
print(f"✅ 데이터 생성 완료! 총 {len(df_orders)}개의 주문 데이터가 '{output_filename}' 파일로 저장되었습니다.")
print(df_orders.head())
```
