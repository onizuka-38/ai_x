```python
import pandas as pd
import glob
import os
```

```python
# 엑셀 폴더 경로

folder_path = "data/"
```

```python
# 엑셀 파일 전부 열기

excel_files = glob.glob(os.path.join(folder_path, "*.xlsx"))
excel_files = [f for f in excel_files if not os.path.basename(f).startswith("~$")]
```

```python
# 통합 데이터 저장 리스트

df_list = []
```

```python
# 각 파일 반복 처리

for file in excel_files :
    data = pd.read_excel(file, sheet_name = None)
    first_sheet_name = list(data.keys())[0]
    df = data[first_sheet_name]
    trimmed_df = df.iloc[1:-3].copy()
    trimmed_df.columns = ["날짜", "가격", "등락률"]
    df_list.append(trimmed_df)
```

```python
df_list
```

```python
# 통합 및 중복 제거

final_df = pd.concat(df_list, ignore_index = True)
final_df = final_df.drop_duplicates()  # 중복 행 제거
final_df = final_df.iloc[1:].reset_index(drop = True)
final_df.at[0, "등락률"] = 0
```

```python
final_df
```

```python
# 저장 파일 설정 및 경로 설정

folder_name = "저장할 폴더 경로"
file_name = "파일이름"
output_path = os.path.join(folder_name, file_name)
```

```python
# 엑셀 저장

final_df.to_excel(output_path, index = False)
```
